````python
# DESIGN FIX: this cell used to cap torque_nm at +-1.5*STD via treat_outliers(method='STD').
# But aula 1 showed these "outliers" are not noise - they ARE the failure signal for
# power_f (95 cases) and overstrain_f (98 cases), both defined by torque sitting far
# outside the typical range. Capping silently deleted that signal from every training
# example before the model ever saw it: mec_power_w's real range (1148-10470 W) collapsed
# to 4367-8571 W after the cap, swallowing the entire [<3500, >9000] failure zone that
# aula 1 proved is 100% deterministic. The model was never shown a single power failure.
#
# We keep the exploratory plot below for visibility (it's still useful to *look* at the
# spread), but we do NOT cap torque_nm: what statistics calls an outlier here is physics
# calling a failure.
renamed_treat_df = renamed_df.copy()

# Reuses the physical variable built in Lesson 1: P = torque x angular velocity.
renamed_treat_df["mec_power_w"] = (
    renamed_treat_df["torque_nm"] * renamed_treat_df["speed_rpm"] * 2 * np.pi / 60
)

columns = ["mec_power_w"]

methods.plot_outliers("after (sem cap - extremos sao sinal, nao ruido)", renamed_treat_df, columns, plt, sns)
````

## The machine failure consists of five independent failure modes:

**1. Tool Wear Failure (tool_wear_f):** the tool will be replaced or fail at a randomly selected tool wear time between 200 and 240 mins (120 times in our dataset). At this point in time, the tool is replaced 69 times, and fails 51 times (randomly assigned).

**2. Heat Dissipation Failure (heat_diss_f):** heat dissipation causes a process failure, if the difference between air- and process temperature is below 8.6 K and the tool rotational speed is below 1380 rpm. This is the case for 115 data points.

**3. Power Failure (power_f):** the product of torque and rotational speed (in rad/s) equals the power required for the process. If this power is below 3500 W or above 9000 W, the process fails, which is the case 95 times in our dataset.

**4. Overstrain Failure (overstrain_f):** if the product of tool wear and torque exceeds 11,000 min Nm for the L product variant (12,000 M, 13,000 H), the process fails due to overstrain. This is true for 98 datapoints.

**5. Random Failures (random_f):** each process has a chance of 0.1% to fail regardless of its process parameters. This is the case for only 5 datapoints, less than could be expected for 10,000 datapoints in our dataset.

````python
features = ["air_temp_k", "process_temp_k", "speed_rpm", "torque_nm", "tool_wear_min", "mec_power_w"]

# Declare feature vector
X = renamed_treat_df[features]

# 	Machine failure	ToolWear	HeatDiss	Power	Overstrain	Random

# Declare target variable
y = renamed_treat_df['failures']
#y = renamed_treat_df['tool_wear_f']  # tool wear
#y = renamed_treat_df['heat_diss_f'] # heat dissipation
#y = renamed_treat_df['power_f'] # power failure
#y = renamed_treat_df['overstrain_f'] # overstrain or excessive torque

# split the dataset into the training set and test set
# IMPORTANT: Stratify by Type to maintain L/M/H distribution (60%/30%/10%) in both train and test sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0, stratify=renamed_treat_df['Type'])

# Verify stratification worked correctly
print("Original Type distribution:")
print(renamed_treat_df['Type'].value_counts(normalize=True).sort_index())

print("\nTraining set Type distribution:")
print(renamed_treat_df.loc[X_train.index, 'Type'].value_counts(normalize=True).sort_index())

print("\nTest set Type distribution:")
print(renamed_treat_df.loc[X_test.index, 'Type'].value_counts(normalize=True).sort_index())
````

````python
import numpy as np
from sklearn.metrics import (average_precision_score, roc_auc_score,
                             precision_recall_curve, classification_report)

neg, pos = np.bincount(y_train)[[0, 1]]

clf = lgb.LGBMClassifier(
    learning_rate=0.05,
    n_estimators=500,
    num_leaves=15,
    max_bin=125,
    min_child_samples=10,
    scale_pos_weight=neg / pos,   # ≈ 29
    importance_type='gain',
    random_state=42,
    verbose=-1,                   # silences the log
)
clf.fit(X_train, y_train)

proba = clf.predict_proba(X_test)[:, 1]
print("PR-AUC :", average_precision_score(y_test, proba))
print("ROC-AUC:", roc_auc_score(y_test, proba))
````

````python
# Safety-Critical Metrics Analysis
from sklearn.metrics import recall_score, precision_score, fbeta_score, f1_score

# Use the explicit confusion-matrix decomposition to avoid swapping TN/TP or FP/FN.
tn, fp, fn, tp = cm.ravel()

# Calculate key metrics
recall = recall_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
miss_rate = fn / (fn + tp) if (fn + tp) > 0 else 0
false_positive_rate = fp / (tn + fp) if (tn + fp) > 0 else 0
f2_score = fbeta_score(y_test, y_pred, beta=2)

# Display results
print("=" * 60)
print("SAFETY-CRITICAL METRICS ANALYSIS")
print("=" * 60)
print(f"\nPRIMARY SAFETY METRICS:")
print(f"   Recall (Sensitivity):        {recall:.2%}  <- Detection Rate")
print(f"   Miss Rate (1 - Recall):      {miss_rate:.2%}  <- SAFETY RISK")
print(f"\nSUPPORTING METRICS:")
print(f"   Specificity:                 {specificity:.2%}  <- Safe Detection")
print(f"   False Positive Rate:         {false_positive_rate:.2%}  <- False Alarms")
print(f"   Precision:                   {precision:.2%}  <- Prediction Accuracy")
print(f"\nCOMPOSITE SCORES:")
print(f"   F1 Score (balanced):         {f1_score(y_test, y_pred):.4f}")
print(f"   F2 Score (safety-weighted):  {f2_score:.4f}  <- Preferred when recall matters")
print("=" * 60)

# Cost Analysis
cost_fn = 1_000_000  # $1M per missed failure
cost_fp = 10_000     # $10K per false alarm
fn_count = fn
fp_count = fp

total_cost = (fn_count * cost_fn) + (fp_count * cost_fp)

print(f"\nCOST ANALYSIS:")
print(f"   False Negatives: {fn_count:3d} x ${cost_fn:,} = ${fn_count * cost_fn:,}")
print(f"   False Positives: {fp_count:3d} x ${cost_fp:,}   = ${fp_count * cost_fp:,}")
print(f"   {'-' * 50}")
print(f"   Total Expected Cost:              ${total_cost:,}")
print("=" * 60)

# Safety Assessment
print(f"\nAVIATION SAFETY ASSESSMENT:")
if recall < 0.90:
    print(f"   INSUFFICIENT: {recall:.1%} detection rate is too low")
    print(f"      Target: >95% for safety-critical systems")
    print(f"      Action: Adjust decision threshold or retrain model")
elif recall < 0.95:
    print(f"   MARGINAL: {recall:.1%} detection rate is borderline")
    print(f"      Consider improvements to reach >95% target")
else:
    print(f"   ACCEPTABLE: {recall:.1%} detection rate meets safety standards")

print("=" * 60)
````

## Safety-Critical Metrics Analysis for Aviation Maintenance

For aviation applications, we must prioritize **false negative reduction** because missed failures are catastrophic. 

See **CONFUSION_MATRIX_GUIDE.md** for detailed explanation of why these metrics matter.

## Sistema de falhas aeronáuticas — decisão GO / NO-GO

O bloco anterior mostrou que, no limiar padrão (0,5), o modelo fica **abaixo da meta de segurança** definida no guia (recall ≥ 95%). Um modelo de classificação sozinho não é "o sistema" — o sistema é a combinação **modelo + limiar de decisão + regra de veredito**, porque é o limiar que decide o trade-off entre falhas perdidas (FN) e alarmes falsos (FP).

Esta seção:
1. varre limiares de decisão e encontra o **menor limiar que garante recall ≥ 95%** (mais alarmes falsos, mas detecta quase todas as falhas reais);
2. compara esse limiar contra o padrão (0,5) e contra o ótimo de F2, replicando a tabela do Step 6 do guia;
3. empacota o resultado numa função `aviation_failure_system(...)` que devolve um veredito **GO** (liberar operação) / **NO-GO** (inspecionar) por registro, igual a um sistema real de bordo tomaria.

````python
from sklearn.metrics import precision_recall_curve

SAFETY_RECALL_TARGET = 0.95  # meta do guia para sistemas de aviacao

def best_threshold_for_recall(y_true, y_proba, target_recall=SAFETY_RECALL_TARGET):
    """Menor limiar (mais restritivo possivel) que ainda garante recall >= target_recall."""
    precision, recall, thresholds = precision_recall_curve(y_true, y_proba)
    # precision/recall tem 1 elemento a mais que thresholds (ultimo ponto = limiar infinito)
    candidates = np.where(recall[:-1] >= target_recall)[0]
    if len(candidates) == 0:
        return 0.0  # meta inalcancavel: sinaliza "classifique tudo como falha"
    return float(thresholds[candidates[np.argmax(thresholds[candidates])]])


def best_threshold_for_f2(y_true, y_proba, grid=np.linspace(0.01, 0.99, 99)):
    best_t, best_f2 = 0.5, -1.0
    for t in grid:
        pred = (y_proba >= t).astype(int)
        if pred.sum() == 0:
            continue
        f2 = fbeta_score(y_true, pred, beta=2)
        if f2 > best_f2:
            best_t, best_f2 = t, f2
    return best_t


thr_safety = best_threshold_for_recall(y_test, y_proba)
thr_f2 = best_threshold_for_f2(y_test, y_proba)

print(f"Limiar padrao                         : 0.500")
print(f"Limiar otimo de F2                    : {thr_f2:.3f}")
print(f"Limiar minimo p/ recall >= {int(SAFETY_RECALL_TARGET*100)}%        : {thr_safety:.4f}")
````

````python
def evaluate_threshold(name, y_true, y_proba, threshold):
    pred = (y_proba >= threshold).astype(int)
    cm_t = confusion_matrix(y_true, pred, labels=[0, 1])
    tn_t, fp_t, fn_t, tp_t = cm_t.ravel()
    return {
        "Cenario": name,
        "Limiar": round(threshold, 4),
        "Recall": recall_score(y_true, pred),
        "Precisao": precision_score(y_true, pred, zero_division=0),
        "FPR": fp_t / (tn_t + fp_t) if (tn_t + fp_t) else 0.0,
        "F2": fbeta_score(y_true, pred, beta=2),
        "FN": fn_t,
        "FP": fp_t,
        "Custo_USD": fn_t * cost_fn + fp_t * cost_fp,
    }

comparison = pd.DataFrame([
    evaluate_threshold("Padrao (0.5)", y_test, y_proba, 0.5),
    evaluate_threshold("Otimo F2", y_test, y_proba, thr_f2),
    evaluate_threshold(f"Seguranca (recall>={int(SAFETY_RECALL_TARGET*100)}%)", y_test, y_proba, thr_safety),
])

fmt = comparison.copy()
fmt["Recall"] = fmt["Recall"].map("{:.2%}".format)
fmt["Precisao"] = fmt["Precisao"].map("{:.2%}".format)
fmt["FPR"] = fmt["FPR"].map("{:.2%}".format)
fmt["F2"] = fmt["F2"].map("{:.3f}".format)
fmt["Custo_USD"] = fmt["Custo_USD"].map("${:,.0f}".format)
print(fmt.to_string(index=False))
````

### Leitura da tabela

Bater a meta de recall ≥ 95% tem preço: o limiar precisa cair muito, o que dispara uma quantidade grande de alarmes falsos (FP sobe, precisão despenca). Isso **não é um defeito da análise — é o trade-off real do problema**, e é exatamente o que o guia descreve na frase final:

> "In aviation, we prefer 100 false alarms over 1 missed failure!"

Ainda assim, o custo total esperado do cenário "Segurança" tende a ser menor que o do cenário padrão, porque cada falha perdida custa 100x mais que um alarme falso (US$ 1.000.000 vs US$ 10.000, conforme `CONFUSION_MATRIX_GUIDE.md`). A tabela acima é o argumento quantitativo para justificar esse limiar mais agressivo perante quem só olha "quantos alarmes falsos vocês geram".

````python
def aviation_failure_system(model, X, threshold=thr_safety):
    """
    Sistema de decisao GO / NO-GO para manutencao preditiva aeronautica.

    Parametros
    ----------
    model     : classificador treinado com .predict_proba
    X         : DataFrame de features (mesmas colunas usadas no treino)
    threshold : limiar de decisao (default: o limiar calibrado para recall >= 95%)

    Retorna
    -------
    DataFrame com a probabilidade de falha e o veredito por registro.
    """
    proba = model.predict_proba(X)[:, 1]
    verdict = np.where(proba >= threshold, "NO-GO — inspecionar", "GO — liberar operacao")
    return pd.DataFrame({
        "failure_probability": proba,
        "threshold": threshold,
        "verdict": verdict,
    }, index=X.index)


# Aplica o sistema ao conjunto de teste inteiro
system_output = aviation_failure_system(clf, X_test, threshold=thr_safety)

print(system_output["verdict"].value_counts().to_string())
print()
system_output.head(10)
````

### O que este sistema é — e o que ele não é

- **É** uma regra de decisão calibrada e auditável: dado um registro, devolve `GO` ou `NO-GO` com o limiar explícito que a produziu, e esse limiar foi escolhido para atingir a meta de recall que a própria aviação exige (≥95%), não por padrão arbitrário (0,5).
- **Não é** um substituto para julgamento de engenharia: o alto número de `NO-GO` no limiar de segurança é o preço documentado do recall alto — precisa ser comunicado a quem opera a manutenção, com a tabela de custo como justificativa, exatamente como o guia recomenda no "Step 4: Cost Analysis".
- Assim como na aula 1, vale a mesma lição: cada modo de falha determinístico (`power_f`, `heat_diss_f`, `overstrain_f`) já é resolvido por uma regra de três linhas, sem necessidade de ML. Este sistema de ML só se justifica para o resíduo não determinístico (`tool_wear_f`, `random_f`) — e mesmo assim, o limiar precisa ser escolhido em função do custo real de um FN vs. um FP, não da acurácia.

## Exportar o modelo para uso offline

Tudo acima depende do kernel do Jupyter estar de pé. Para usar o sistema fora do notebook — num
script, num serviço, ou só para não re-treinar toda vez — salvamos o classificador treinado e a
configuração de decisão (features, limiares, custos) em disco.

````python
import json
from pathlib import Path
import joblib

MODEL_DIR = Path("model")
MODEL_DIR.mkdir(exist_ok=True)

# Modelo completo (wrapper scikit-learn: preserva predict_proba, feature_names_, etc.)
joblib.dump(clf, MODEL_DIR / "lgbm_aviation_model.pkl")

# Booster nativo tambem, em formato texto - le em qualquer binding do LightGBM (C++, R, Python puro)
clf.booster_.save_model(str(MODEL_DIR / "lgbm_aviation_model.txt"))

# Configuracao de decisao: sem isso, o .pkl sozinho nao sabe qual limiar usar nem o custo de cada erro
config = {
    "features": features,
    "thresholds": {
        "default": 0.5,
        "f2_optimal": round(thr_f2, 6),
        "safety_recall_95": round(thr_safety, 6),
    },
    "recommended_threshold": round(thr_safety, 6),
    "costs_usd": {"false_negative": cost_fn, "false_positive": cost_fp},
    "test_set_metrics_at_recommended_threshold": {
        "recall": recall_score(y_test, (y_proba >= thr_safety).astype(int)),
        "precision": precision_score(y_test, (y_proba >= thr_safety).astype(int), zero_division=0),
        "f2": fbeta_score(y_test, (y_proba >= thr_safety).astype(int), beta=2),
    },
}
with open(MODEL_DIR / "model_config.json", "w") as f:
    json.dump(config, f, indent=2)

print("Salvo em:", MODEL_DIR.resolve())
for p in sorted(MODEL_DIR.iterdir()):
    print(" -", p.name, f"({p.stat().st_size:,} bytes)")
````

Isso gera `model/lgbm_aviation_model.pkl` (o classificador completo), `model/lgbm_aviation_model.txt`
(o booster nativo do LightGBM, legível por qualquer binding — R, C++, Python puro) e
`model/model_config.json` (features, limiares e custos). Para usar sem o Jupyter, veja
`predict_offline.py` na mesma pasta:

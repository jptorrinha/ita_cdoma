## Modelo fornecido — execute sem alterar

A célula seguinte define a POD, o simulador de uma frota e a função que executa as réplicas. O fator de crescimento multiplica a distribuição do coeficiente `A` na inicialização e após cada renovação. A função `repetir_estrutura` devolve os vetores de falhas e horas, com um valor por réplica.


```python
def pod(tamanho_mm, beta=1.5, eta=10.0, gama=1.0):
    x = np.maximum(np.asarray(tamanho_mm, dtype=float) - gama, 0.0)
    return 1.0 - np.exp(-np.power(x / eta, beta))


def simular_estruturas(rng, n=1_000, limiar=15_000, intervalo=7_500, vida_fc=80_000, fator_crescimento=1.0):
    fh_por_fc = rng.uniform(1.3, 1.7, size=n)
    a0 = rng.uniform(0.1, 1.0, size=n)
    A = fator_crescimento * rng.uniform(1.8e-4, 2.0e-4, size=n)
    fc = np.zeros(n)
    rep = np.zeros(n, dtype=int)
    ativo = np.ones(n, dtype=bool)
    falhas = 0
    while ativo.any():
        idx = np.flatnonzero(ativo)
        # Antes do primeiro reparo, `prox` é a idade absoluta da aeronave; depois, o intervalo desde o reparo.
        prox = (limiar + intervalo * rep[idx] + rng.integers(-1000, 1, size=idx.size)).astype(float)
        fim = (fc[idx] + prox) >= vida_fc
        prox[fim] = vida_fc - fc[idx][fim]
        a = a0[idx] * np.exp(A[idx] * prox)
        falhou = a >= 100.0
        w = idx[falhou]
        fc[w] += np.ceil(np.log(100.0 / a0[w]) / A[w])
        falhas += w.size
        a0[w], A[w], rep[w] = rng.uniform(0.1, 1.0, w.size), fator_crescimento * rng.uniform(1.8e-4, 2.0e-4, w.size), -1
        det = (rng.random(idx.size) < pod(a)) & ~falhou & ~fim
        w = idx[det]
        fc[w] += prox[det]
        a0[w], A[w], rep[w] = rng.uniform(0.1, 1.0, w.size), fator_crescimento * rng.uniform(1.8e-4, 2.0e-4, w.size), -1
        rep[idx] += 1
        ativo[idx[fim & ~falhou]] = False
    return falhas, float(np.sum(vida_fc * fh_por_fc))   # denominador nominal


def repetir_estrutura(limiar, intervalo, n_rep, semente, fator_crescimento=1.0):
    rng = np.random.default_rng(semente)
    r = [simular_estruturas(rng, limiar=limiar, intervalo=intervalo, fator_crescimento=fator_crescimento) for _ in range(n_rep)]
    return np.array([f for f, _ in r]), np.array([h for _, h in r])

```

A próxima célula já contém os parâmetros e sementes exigidos. Execute-a e aguarde a conclusão dos três cenários. Depois, complete apenas a célula de cálculo dos indicadores e apresentação da tabela.


```python
N_REP = 10_000
CRITERIO = 1e-9
# nome, primeira inspeção (ciclos), intervalo (ciclos), fator de crescimento, semente
CENARIOS = [
    ("Base", 15_000, 7_500, 1.0, 2026),
    ("Revisto", 10_000, 5_000, 1.0, 2027),
    ("Revisto: crescimento +20%", 10_000, 5_000, 1.2, 2028),
]
resultados = {}
for nome, limiar, intervalo, fator, semente in CENARIOS:
    falhas, horas = repetir_estrutura(limiar, intervalo, N_REP, semente, fator)
    resultados[nome] = {"falhas": falhas, "horas": horas, "semente": semente}
    print(nome, "— execução concluída:", len(falhas), "réplicas")
```

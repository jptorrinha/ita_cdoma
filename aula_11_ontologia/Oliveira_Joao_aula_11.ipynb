## Modelo fornecido — execute sem alterar

A célula seguinte define a POD, o simulador de uma frota e a função que executa as réplicas. O fator de crescimento multiplica a distribuição do coeficiente `A` na inicialização e após cada renovação. A função `repetir_estrutura` devolve os vetores de falhas e horas, com um valor por réplica.


```python
def pod(tamanho_mm, beta=1.5, eta=10.0, gama=1.0):
    x = np.maximum(np.asarray(tamanho_mm, dtype=float) - gama, 0.0)
    return 1.0 - np.exp(-np.power(x / eta, beta))


def simular_estruturas(rng, n=1_000, limiar=15_000, intervalo=7_500, vida_fc=80_000, fator_crescimento=1.0):
    fh_por_fc = rng.uniform(1.3, 1.7, size=n)
    a0 = rng.uniform(0.1, 1.0, size=n)
    A = fator_crescimento * rng.uniform(1.8e-4, 2.0e-4, size=n)
    fc = np.zeros(n)
    rep = np.zeros(n, dtype=int)
    ativo = np.ones(n, dtype=bool)
    falhas = 0
    while ativo.any():
        idx = np.flatnonzero(ativo)
        # Antes do primeiro reparo, `prox` é a idade absoluta da aeronave; depois, o intervalo desde o reparo.
        prox = (limiar + intervalo * rep[idx] + rng.integers(-1000, 1, size=idx.size)).astype(float)
        fim = (fc[idx] + prox) >= vida_fc
        prox[fim] = vida_fc - fc[idx][fim]
        a = a0[idx] * np.exp(A[idx] * prox)
        falhou = a >= 100.0
        w = idx[falhou]
        fc[w] += np.ceil(np.log(100.0 / a0[w]) / A[w])
        falhas += w.size
        a0[w], A[w], rep[w] = rng.uniform(0.1, 1.0, w.size), fator_crescimento * rng.uniform(1.8e-4, 2.0e-4, w.size), -1
        det = (rng.random(idx.size) < pod(a)) & ~falhou & ~fim
        w = idx[det]
        fc[w] += prox[det]
        a0[w], A[w], rep[w] = rng.uniform(0.1, 1.0, w.size), fator_crescimento * rng.uniform(1.8e-4, 2.0e-4, w.size), -1
        rep[idx] += 1
        ativo[idx[fim & ~falhou]] = False
    return falhas, float(np.sum(vida_fc * fh_por_fc))   # denominador nominal


def repetir_estrutura(limiar, intervalo, n_rep, semente, fator_crescimento=1.0):
    rng = np.random.default_rng(semente)
    r = [simular_estruturas(rng, limiar=limiar, intervalo=intervalo, fator_crescimento=fator_crescimento) for _ in range(n_rep)]
    return np.array([f for f, _ in r]), np.array([h for _, h in r])

```

## Partes 1 e 2 — Execute os três cenários

A próxima célula já contém os parâmetros e sementes exigidos. Execute-a e aguarde a conclusão dos três cenários. Depois, complete apenas a célula de cálculo dos indicadores e apresentação da tabela.


```python
N_REP = 10_000
CRITERIO = 1e-9
# nome, primeira inspeção (ciclos), intervalo (ciclos), fator de crescimento, semente
CENARIOS = [
    ("Base", 15_000, 7_500, 1.0, 2026),
    ("Revisto", 10_000, 5_000, 1.0, 2027),
    ("Revisto: crescimento +20%", 10_000, 5_000, 1.2, 2028),
]
resultados = {}
for nome, limiar, intervalo, fator, semente in CENARIOS:
    falhas, horas = repetir_estrutura(limiar, intervalo, N_REP, semente, fator)
    resultados[nome] = {"falhas": falhas, "horas": horas, "semente": semente}
    print(nome, "— execução concluída:", len(falhas), "réplicas")
```

```python
# Indicadores e tabela dos três cenários.
def erro_padrao(vetor):
    return vetor.std(ddof=1) / np.sqrt(vetor.size)


linhas = []
for nome, dados in resultados.items():
    falhas = dados["falhas"]
    horas = dados["horas"]
    taxas = falhas / horas
    taxa_media, ep_taxa = taxas.mean(), erro_padrao(taxas)
    linhas.append({
        "cenario": nome,
        "n": falhas.size,
        "semente": dados["semente"],
        "falhas_media": falhas.mean(),
        "ep_falhas": erro_padrao(falhas),
        "taxa_media": taxa_media,
        "ep_taxa": ep_taxa,
        "lim_inf": taxa_media - 2 * ep_taxa,
        "lim_sup": taxa_media + 2 * ep_taxa,
        "todas_zero": bool((falhas == 0).all()),
    })

cab = (f"{'Cenário':<27} {'n':>6} {'Semente':>7} {'Falhas médias':>13} {'EP falhas':>9} "
       f"{'Taxa média (/h)':>15} {'EP taxa (/h)':>12} {'Lim. inf. (/h)':>14} {'Lim. sup. (/h)':>14}")
print(cab)
print("-" * len(cab))
for l in linhas:
    print(f"{l['cenario']:<27} {l['n']:>6,d} {l['semente']:>7d} {l['falhas_media']:>13.4f} {l['ep_falhas']:>9.4f} "
          f"{l['taxa_media']:>15.3e} {l['ep_taxa']:>12.3e} {l['lim_inf']:>14.3e} {l['lim_sup']:>14.3e}")

# Leitura de cada faixa de dois EP pela regra didática do enunciado
print(f"\nCritério didático: {CRITERIO:.0e} /h")
for l in linhas:
    if l["todas_zero"]:
        leitura = "todas as réplicas com zero falhas: a amostra não permite concluir pela ausência de risco"
    elif l["lim_sup"] < CRITERIO:
        leitura = "faixa inteiramente abaixo do critério"
    elif l["lim_inf"] > CRITERIO:
        leitura = "faixa inteiramente acima do critério"
    else:
        leitura = "faixa inclui o critério: inconclusivo com a precisão obtida"
    print(f"{l['cenario']:<27} [{l['lim_inf']:.3e}; {l['lim_sup']:.3e}] -> {leitura}")
```

## Parte 3 — Interpretar os limites

**1. Exposição e denominador nominal.** Explique o significado e a unidade da soma de 80.000 ciclos × horas por ciclo de cada aeronave, seu uso no cálculo da taxa e uma situação em que a exposição operacional real poderia ser diferente. Não altere o código.

*Resposta:* A soma de 80.000 ciclos × horas por ciclo de cada aeronave é a **exposição nominal da frota**: o total de horas de voo que as 1.000 aeronaves acumulariam se cada uma completasse a vida de 80.000 ciclos. A unidade é hora porque ciclos × (horas/ciclo) = horas; no código, as horas por ciclo são sorteadas entre 1,3 e 1,7 por aeronave, e a soma fica perto de 1,2 × 10⁸ h por réplica. Ela é o **denominador** da taxa: em cada réplica, taxa = falhas estruturais ÷ horas de exposição, e a tabela usa a média dessas taxas por réplica. Por ser nominal, ela não depende do que acontece na simulação: uma aeronave que falha é renovada e continua contando até 80.000 ciclos. Na operação real, a exposição poderia ser menor ou diferente quando aeronaves são retiradas de serviço antes do fim da vida, ficam paradas por longos períodos de manutenção ou indisponibilidade, ou são perdidas após uma falha estrutural. Também mudaria se o perfil de missão alterar as horas por ciclo, por exemplo com voos de treinamento curtos e muitos pousos. Contar menos horas do que as efetivamente voadas aumenta a taxa sem nenhuma falha adicional, e vice-versa.
